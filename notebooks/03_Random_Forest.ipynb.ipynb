```python
import joblib

X_train = joblib.load("X_train.pkl")
X_test = joblib.load("X_test.pkl")
y_train = joblib.load("y_train.pkl")
y_test = joblib.load("y_test.pkl")

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
y_pred_rf = rf_model.predict(X_test)

print(y_pred_rf[:10])
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_rf))
```

```python
from sklearn.metrics import confusion_matrix

cm_rf = confusion_matrix(y_test, y_pred_rf)

print(cm_rf)
```

```python
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

print(y_prob_rf[:10])
```

```python
from sklearn.metrics import roc_auc_score

auc_rf = roc_auc_score(y_test, y_prob_rf)

print("Random Forest ROC-AUC:", auc_rf)
```

```python
import pandas as pd

feature_importance = pd.DataFrame({
    'Feature': X_train.columns,
    'Importance': rf_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    by='Importance',
    ascending=False
)

feature_importance.head(10)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_model2 = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    min_samples_split=10,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model2.fit(X_train, y_train)

print("Improved RF trained!")
```

```python
y_prob_rf2 = rf_model2.predict_proba(X_test)[:,1]

print(y_prob_rf2[:10])
```

```python
y_pred_rf2 = rf_model2.predict(X_test)

print(y_pred_rf2[:10])
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_rf2))
```

```python
from sklearn.metrics import roc_auc_score

auc_rf2 = roc_auc_score(y_test, y_prob_rf2)

print("Improved RF ROC-AUC:", auc_rf2)
```

```python
import joblib

joblib.dump(rf_model2, "final_random_forest_model.pkl")

print("Model saved successfully!")
```

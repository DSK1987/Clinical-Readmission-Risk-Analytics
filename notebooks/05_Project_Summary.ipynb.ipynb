This project develops a machine learning model to predict hospital readmissions using patient demographic, clinical, and healthcare utilization data.

The objective is to identify patients at high risk of readmission and support early clinical intervention.

# 1. Problem Statement

Hospital readmissions are a major challenge for healthcare systems worldwide.

Frequent readmissions increase healthcare costs, place pressure on hospital resources, and may indicate inadequate disease management or discharge planning.

The aim of this project was to develop a machine learning model capable of predicting whether a patient is likely to be readmitted based on information available during hospitalization.

Early identification of high-risk patients may support clinicians in implementing targeted interventions and improving patient outcomes.

# 2. Dataset Description

The dataset contains hospital encounter records for diabetic patients.

Key characteristics:

- Total records: 101,766
- Target variable: Readmission status
- Binary outcome:
  - 0 = No readmission
  - 1 = Readmission

Variables included:

- Demographics
  - Age
  - Gender
  - Race

- Admission information
  - Admission type
  - Admission source
  - Discharge disposition

- Clinical information
  - Diagnoses
  - Medications
  - Laboratory procedures

- Healthcare utilization
  - Previous inpatient visits
  - Previous outpatient visits
  - Emergency visits
 
# 3. Data Preprocessing

Several preprocessing steps were performed before model development.

### Missing Value Handling

Missing values in categorical variables were replaced with meaningful categories.

### Feature Engineering

Diagnosis codes were grouped into broader clinical categories:

- Diabetes
- Circulatory diseases
- Respiratory diseases
- Digestive diseases
- Injury
- Other

### Target Variable Creation

Readmission status was converted into a binary variable.

### Categorical Encoding

One-hot encoding was applied to categorical variables.

### Train-Test Split

The dataset was split into:

- 80% Training Set
- 20% Test Set

Stratified sampling was used to preserve class distribution.

# 4. Model Development

Two machine learning models were developed and evaluated.

## Logistic Regression

A balanced logistic regression model was trained to address class imbalance.

## Random Forest

A Random Forest classifier was developed and further optimized using:

- 300 decision trees
- Maximum depth = 10
- Balanced class weights

Random Forest was selected as the final model due to superior performance.

# 5. Model Performance

```python
import pandas as pd

results = pd.DataFrame({
    'Model': ['Balanced Logistic Regression',
              'Improved Random Forest'],
    'ROC-AUC': [0.652, 0.664],
    'Recall (Readmission)': [0.55, 0.55],
    'Accuracy': [0.65, 0.67]
})

results
```

# 6. ROC Curve

The ROC curve evaluates the ability of the model to distinguish between readmitted and non-readmitted patients.

The Random Forest model achieved:

ROC-AUC = 0.664

This indicates moderate discrimination ability and performance better than random classification.
